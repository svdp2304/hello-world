```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create sample data
data = pd.DataFrame({
    'x': np.arange(10),
    'y': np.random.randn(10)
})
```

```python
data.head()
```

```python
# Basic plot
sns.lineplot(x='x', y='y', data=data)
plt.title('Hello World Plot')
plt.show()
```
